```python
# generate data
P = 10

x = np.linspace(0, 1, 100)
y = np.sin(x*10)

X = x[:, None]
Y = y[:, None]

print(X.shape, Y.shape)
```

```python
YP = np.hstack([Y+p for p in range(P)])
```

```python
# Construct model

m_prior_list = objax.ModuleList([
    lego.models.GP(X)
    for p in range(P)
])

m_post_list = objax.ModuleList([
    lego.models.GP(X, Y+p)
    for p in range(P)
])

multi_output_gp = lego.models.GP(X, YP)
```

```python
def get_objective(y_p, m, m2):
    return jnp.sum(y_p), [m.get_objective(), 1]
```

```python
%%timeit -r 1 -n 1
# Manual 
a_list = []
b_list = []
for p in range(P):
    a, b = get_objective(YP[:, p], m_post_list[p], m_post_list[p])
    a_list.append(a)
    b_list.append(b)
print(np.array(a_list))
print(np.array(b_list))
```

```python
%%timeit -r 1 -n 1
# vmap approach
a_val, b_val = batchjax.batch_fn(
    get_objective, 
    [YP, m_post_list, m_post_list], 
    axes=[1, 0, 0], 
    out_dim=2
)
print(a_val)
print(b_val)
```

```python
%%timeit -r 1 -n 1
a_val, b_val = batchjax.batch_or_loop(
    get_objective, 
    [YP, m_post_list, m_post_list], 
    axes=[1, 0, 0], 
    dim = P,
    out_dim=2,
    batch_flag=False
)
print(a_val)
print(b_val)
```
