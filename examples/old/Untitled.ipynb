```python
import numpy as np
```

```python
# in batched latent-data format
A = np.array([
    [1, 2, 3],
    [4, 5, 6], 
])
```

```python
# calling reshape on batched latent-data preserves latent-data format
print(A)
print(A.reshape(6))
A_ld_vec = A.reshape(6)t
```

```python
# since reshape preserves the format we can convert back to batched latent-data by reshaping
A_ld_vec.reshape([2, 3])
```

```python
# calling reshape on batched data-latent preserves data-latent format
print(A.T)
print((A.T).reshape(6))
A_dl_vec = (A.T).reshape(6)
```

```python
# since reshape preserves the format we can convert back to batched data-latent by reshaping
A_dl_vec.reshape([3, 2])
```
